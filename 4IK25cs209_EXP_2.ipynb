```python
import numpy as np
import pandas as pd
```

```python
data={
    "name":["Amit","Sup","Chetan","Amit","Diya","Ravi"],
    "age":[30,np.nan, 35, 30,56,54],
    "Marks":[78,86,np.nan,78,98,90],
    "City":["Mysuru","Mysuru","Bengaluru","Mysuru",np.nan,"Delhi"]
}
df=pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\n Shape of dataset(rows, columns):",df.shape)
print("\nColumn names:",df.columns.tolist())
print("\n Data types of each column:")
print(df.dtypes)
print("\nDataset info:")
print(df.info())
```

```python
print("\nDescriptive statistics(numerical columns):")
print(df.describe())
print("\nMean:\n",df.mean(numeric_only=True))
print("\nMedian:\n",df.median(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nmissing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in dataset:", df.isnull().sum().sum())
```

```python
print("\nDuplicate rows in dataset:")
print([df.duplicated()])
print("\nNumber of duplicate rows:",df.duplicated().sum())
print("\nEDA process completed sucessfully!")
```
