```python
import numpy as np
import pandas as pd
print("NumPy version:",np.__version__)
print("Pandas version:",pd.__version__)
```

```python
marks=np.array([78,85,98,90,80])
print("Marks:",marks)
print("Type:",type(marks))
```

```python
print("Number of dimensions:",marks.ndim)
print("Shape:",marks.shape)
print("Number of elements:",marks.size)
```

```python
marks=np.array([78,85,98,90,80])
print("Original:",marks)
print("Add 5:",marks+5)
print("Multiply by 2:",marks*2)
print("Divide by 2",marks/2)
print("Square:",marks**2)
```

```python
print("Sum:",marks.sum())
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
```

```python
print("First mark:",marks[0])
print("Second mark:",marks[1])
print("Last mark:",marks[2])
```

```python
print("Index 1 to 3:",marks[1:4])
print("First three:",marks[:3])
print("From index 2:",marks[1:4])
print("Every second element:",marks[::2])
```

```python
student_marks=np.array([
    [78,85,92],
    [67,88,76],
    [90,95,89]
])
print(student_marks)
print("Shape:",student_marks.shape)
```

```python
print("Mean of all values:",student_marks.mean())
print("Column wise mean:",student_marks.mean(axis=0))
print("Row wise mean:",student_marks.mean(axis=1))
```

```python
data={
    "Name":["Anu", "Rahul","Kiran","Priya","Asha","Bharath"],
    "Department":["CSE","AIML","CSE","AIML","ISE","CSe"],
    "Marks":[78,89,90,98,78,84],
    "Attendance":[97,78,87,86,77,67]
}
st=pd.DataFrame(data)
st
```

```python
print("\nName and Marks:")
print(st[["Name","Marks"]])
print("\nStudents scoring 80 or above:")
display(st[st["Marks"]>=80])


print("\nStudents having attendance above 85%:")
display(st[st["Attendance"]>=85])

print("First three rows:")
display(st.iloc[0:3])

print([st["Marks"].mean()])

print([st["Marks"].max()])

print("Students from CSE:")
display(st[st["Department"]=="CSE"])

print("\nStudents scoring 80 or above with attedance 85%:")
display(st[
        (st["Marks"]>=80) &
        (st["Attendance"]>85)]
)







```
