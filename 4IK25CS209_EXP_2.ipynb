```python
import pandas as pd
import numpy as np
```

Step1 : Create /Load the Dataset

```python
data ={
    'Name' : ['Amit', 'Bebo' ,'Chetan' , 'Amit','Divya', 'Ravi', 'Sneha' , 'Karan', 'Pooja',
              'Manish' ,'Anjali', 'Suresh' , 'Kavya' ,'Rahul ','Meera' , 'Vikram' ,'Nisha',
              'Arjun' ,'Divya','Rohit','Swathi','Naveen','Priya','Karan','Deepa'],

    'Age':[25,np.nan,30,25,28,22,24,np.nan,27,26,23,29,np.nan,25,22,28,24,np.nan,28,
           26,23,30,27,np.nan,25],

    'Marks':[85,90,np.nan,85,78,95,88,72,np.nan,91,67,84,79,np.nan,93,88,76,82,78,
             np.nan,90,85,71,72,89],

    'City' : ['Mysuru','Mysuru','Bengaluru','Mysuru',np.nan,'Delhi','Mysuru','Mangaluru',
              'Bengaluru',np.nan,'Mysuru','Hubli','Mangaluru','Delhi','Bengaluru',np.nan,
              'Mysuru','Hubli',np.nan,'Delhi','Mangaluru','Mysuru','Bengaluru','Mangaluru',
              'Hubli']
}

df=pd.DataFrame(data)
print("Dataset:")
print(df)
```

Step 2: Basic Information

```python
print("\nShape of Dataset (rows,columns):",df.shape)
```

```python
print("\nColumn names:", df.columns.tolist())
```

```python
print("\n Data types of each column:")
print(df.dtypes)
```

```python
print("\n Data Info:")
df.info()
```

Step 3: Descriptive Statistics

```python
print("\nDescriptive statistis (numerical column):")
print(df.describe())
```

```python
print("\nMean: \n", df.mean(numeric_only=True))
```

```python
print("\nMedian : \n",df.median(numeric_only=True))
```

```python
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

Step 4: Identifying Missing Values

```python
print("\n Missing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in dataset:",df.isnull().sum().sum())
```

Step5 : Identifying Duplicate Values

```python
print("\n Duplicate rows in dataset:")
print(df[df.duplicated()])
print("\n Number of Duplicate rows:", df.duplicated().sum())
print("\n EDA process completed successfully!")
```
